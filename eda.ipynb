```python
import os 
data_dir = os.path.join("data","ESC-50-master")
audio_dir = os.path.join(data_dir, "audio")
label_dir = os.path.join(data_dir, "meta")
```

```python
audio_files = [f for f in os.listdir(audio_dir) if f.endswith('.wav')]
audios_count = len(audio_files) 
```

```python
import pandas as pd
import numpy as np
df = pd.read_csv(os.path.join(label_dir, "esc50.csv"))
df.head(5)
```

```python
df.count()==np.array([audios_count]*len(df.columns)) #checking if the number of audio files is equal to the number of rows in the dataframe for each column
```
